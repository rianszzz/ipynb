### iii. Uji Hipotesis
**Two sampled T-test**: <br>
Uji Independent Samples T Test atau 2-sample T-Test membandingkan rata-rata dua kelompok independen untuk menentukan apakah ada bukti statistik bahwa rata-rata populasi terkait berbeda secara signifikan.


Tentukan null dan alternate hypothesis. Secara umum, null hypothesis akan menyatakan bahwa kedua populasi yang diuji tidak memiliki perbedaan yang signifikan secara statistik. <br><br>
Alternate hypothesis akan menyatakan bahwa ada perbedaan.

    

```python
# Buat uji hipotesis untuk menentukan null atau alternate hypothesis
# CODE HERE
from scipy import stats

group1 = data[data['age'] <= data['age'].median()]['strength']
group2 = data[data['age'] > data['age'].median()]['strength']

t_stat, p_value = stats.ttest_ind(group1, group2)

print('t-statistic:', t_stat)
print('p-value:', p_value)

alpha = 0.05
if p_value < alpha:
    print('Tolak H0: Ada perbedaan signifikan (Alternate hypothesis diterima)')
else:
    print('Gagal tolak H0: Tidak ada perbedaan signifikan (Null hypothesis diterima)')
```

### iv. Kesimpulan akhir

```python
# Berikan kesimpulan yang bisa di ambil dari analysis yg Anda lakukan
print("""
Kesimpulan hasil uji hipotesis:

Uji hipotesis dilakukan menggunakan Independent Samples t-test untuk mengetahui apakah terdapat perbedaan rata-rata kekuatan beton (strength) berdasarkan umur beton (age). Data dibagi menjadi dua kelompok, yaitu beton dengan umur kurang dari atau sama dengan median dan beton dengan umur di atas median.

Hipotesis yang digunakan adalah:

H₀ (Null Hypothesis): Tidak terdapat perbedaan rata-rata kekuatan beton yang signifikan antara kedua kelompok umur beton.

H₁ (Alternative Hypothesis): Terdapat perbedaan rata-rata kekuatan beton yang signifikan antara kedua kelompok umur beton.

Dengan tingkat signifikansi sebesar 5% (α = 0,05 α = 0,05), keputusan ditentukan berdasarkan nilai p-value:

Jika p-value < 0,05, maka H₀ ditolak. Artinya, terdapat perbedaan rata-rata kekuatan beton yang signifikan antara kedua kelompok umur.

Jika p-value ≥ 0,05, maka H₀ gagal ditolak. Artinya, belum terdapat bukti statistik yang cukup untuk menyatakan adanya perbedaan rata-rata kekuatan beton yang signifikan.

Kesimpulan akhir: Hasil pengujian dapat digunakan untuk mengetahui apakah umur beton berhubungan dengan perbedaan kekuatan beton secara statistik. Untuk menentukan apakah H₀ ditolak atau gagal ditolak pada analisis ini, nilai p-value yang dihasilkan dari kode perlu diperhatikan. Secara umum, beton yang berumur lebih lama cenderung memiliki kekuatan lebih tinggi karena proses hidrasi semen berlangsung seiring waktu, tetapi hal tersebut tetap perlu dibuktikan melalui hasil analisis data.
""")
```
